## Costly Signals (Roth-Erev block retired)

The Roth-Erev costly-signaling block was retired on 2026-05-09. Costly signaling is theoretically ill-defined under the project's non-negativity-clamped Roth-Erev urn rule (see `analytics/math/agent_urn.md` "Applicability constraints" and `analytics/math/costly_signaling.md` "Compatibility with the project's three agent types"). Q-learning's TD update is defined on the reals and handles the costly-reward range natively, so the Q-learning costly section below is unaffected.
